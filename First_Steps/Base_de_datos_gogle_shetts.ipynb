```python
import gspread
from oauth2client.service_account import ServiceAccountCredentials

# 1. Configurar la conexión (reemplaza con tu archivo de credenciales de Google Cloud)
scope = ["https://spreadsheets.google.com/feeds", "https://www.googleapis.com/auth/drive"]
creds = ServiceAccountCredentials.from_json_keyfile_name("credenciales_google.json", scope)
client = gspread.authorize(creds)

# 2. "Conectarse" a la Base de Datos (abrir el archivo por su nombre)
base_de_datos = client.open("Household_Items_Valuation")

# 3. Seleccionar la "Tabla" (la pestaña de la hoja)
tabla_items = base_de_datos.worksheet("Items_Valuation")

# --- OPERACIÓN: SELECT * FROM Items_Valuation ---
# Trae todos los registros como una lista de diccionarios
todos_los_datos = tabla_items.get_all_records()
print("Datos actuales:", todos_los_datos)

# --- OPERACIÓN: INSERT INTO Items_Valuation VALUES (...) ---
# Definimos el nuevo registro que queremos guardar
nuevo_item = [
    "Coffee_Maker",       # Google_Vision_Label
    3,                     # Category_ID (Cocina chica)
    20.00,                 # Economy Cost
    60.00,                 # Standard Cost
    250.00                 # Premium Cost
]

# Insertamos la fila al final de la tabla
tabla_items.append_row(nuevo_item)
print("¡Nuevo ítem guardado con éxito!")
```
